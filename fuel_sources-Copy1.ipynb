```python
# To import libraries and load dataset
import pandas as pd
import numpy as np
capacity=pd.read_csv('capacity_csv.csv')
capacity.head()
```

```python
# To check for NaNs and datatype
capacity.info()
```

```python
# To change datatyoes
# Year is already int64, so leave it as is
text_cols = ['State Code', 'Producer Type', 'Fuel Source']
capacity[text_cols] = capacity[text_cols].astype('object')
 
num_cols = ['Generators', 'Facilities',
            'Nameplate Capacity (Megawatts)', 'Summer Capacity (Megawatts)']
 
# Remove thousands separators BEFORE converting, so values like '1,845.0' are kept
for col in num_cols:
    cleaned = capacity[col].astype(str).str.replace(',', '', regex=False).str.strip()
    capacity[col] = pd.to_numeric(cleaned, errors='coerce')
 
capacity[['Generators', 'Facilities']] = capacity[['Generators', 'Facilities']].astype('Int64')
 
capacity.info()
capacity.head()
```

```python
# To drop columns
drop_columns=['Generators','Facilities','Nameplate Capacity (Megawatts)']
capacity=capacity.drop(labels=drop_columns, axis=1)
capacity.head()
```

```python
# To create a function to clean column names and rows
def clean_df(df):
    # To clean column names
    df.columns=df.columns.str.strip().str.lower()

    # To clean whitespace in rows of string datatype columns
    for column in df.columns:
        if df[column].dtype=='int64':
            continue
        elif df[column].dtype=='str':
            df[column]=df[column].str.strip()
        return(df)
```

```python
# To call clean_df function on capacity
capacity=clean_df(capacity)
capacity.head()
```

```python
# To lowercase the rows of columns 3 and 4
capacity['producer type']=capacity['producer type'].str.lower()
capacity['fuel source']=capacity['fuel source'].str.lower()
capacity.head()
```

```python
# To rename column
column_mapper={'summer capacity (megawatts)':'sum_cap(mw)'}
capacity=capacity.rename(mapper=column_mapper,axis=1)
capacity.head()
```

```python
# To rename producer types
capacity['producer type']=capacity['producer type'].str.replace(pat='electric generators, electric utilities',repl='utilities')
capacity['producer type']=capacity['producer type'].str.replace(pat='electric generators, independent power producers',repl='ipp')
capacity.head(25)
```

```python
# Define your list of producers
producers = [
    'utilities', 
    'ipp'
]

# Filter everything in one go
capacity_coal = capacity[
    (capacity['year'] == 2025) & 
    (capacity['fuel source'] == 'coal') & 
    (capacity['producer type'].isin(producers))
]

capacity_coal.head()
```

```python
# Build a lookup of 2025 'all sources' capacity, keyed by state and producer type
all_sources = capacity[
    (capacity['year'] == 2025) &
    (capacity['fuel source'] == 'all sources') &
    (capacity['producer type'].isin(producers))
][['state code', 'producer type', 'sum_cap(mw)']].rename(columns={'sum_cap(mw)': 'all sources'})
 
# Match each coal row to its 'all sources' value
capacity_coal = capacity_coal.merge(all_sources, on=['state code', 'producer type'], how='left')
 
capacity_coal.head()
```

```python
# Coal share of all-source summer capacity, as a percentage
capacity_coal['%'] = (capacity_coal['sum_cap(mw)'] / capacity_coal['all sources'] * 100).round(2)
 
capacity_coal.head()
```

```python
## What states have the highest electric generator capacity uaing coal?
```

```python
# To sort by summer capacity, in descending order, excluding 'US' rows
capacity_coal_sorted_mw = (
    capacity_coal[capacity_coal['state code'] != 'US']
    .sort_values(by='sum_cap(mw)', ascending=False)
)
 
capacity_coal_sorted_mw.head()
```
